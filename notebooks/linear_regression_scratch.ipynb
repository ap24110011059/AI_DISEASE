```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sample data

X = np.array([1, 2, 3, 4, 5])

y = np.array([2, 4, 6, 8, 10])

plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Dataset")
plt.show()
```

```python
m = 0
b = 0

learning_rate = 0.01
epochs = 1000

n = len(X)

cost_history = []
```

```python
for epoch in range(epochs):

    y_pred = m * X + b

    cost = (1 / n) * np.sum((y - y_pred) ** 2)

    cost_history.append(cost)

    dm = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db

print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
plt.scatter(X, y, label="Actual Data")

plt.plot(X, m * X + b, label="Regression Line")

plt.xlabel("X")
plt.ylabel("y")

plt.legend()

plt.show()
```

Linear Regression was implemented using only NumPy without using sklearn.

Gradient Descent was used to update the slope (m) and intercept (b).

The Cost vs Epochs graph shows how the error decreases during training, helping understand how Linear Regression learns from data.
