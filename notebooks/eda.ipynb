```python
columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

df = pd.read_csv("../pima.csv", names=columns)

df.head()
```

```python
df.hist(figsize=(12, 10))
plt.tight_layout()
plt.show()
```

```python
class_counts = df["Outcome"].value_counts()

print(class_counts)

plt.figure(figsize=(5,5))
class_counts.plot(kind="bar")

plt.title("Class Imbalance")
plt.xlabel("Outcome")
plt.ylabel("Count")

plt.xticks([0,1], ["Non-Diabetic", "Diabetic"], rotation=0)

plt.show()
```
